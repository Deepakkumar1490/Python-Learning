```python
# int 
print(10 + 5)
print(10 - 5)
print(10 / 5)
print(10 // 5)
print(10 % 5)
print(10 * 5)
print(10 ** 5)
```

```python
# float
print(10.0 + 5.0)
print(10.0 - 5.0)
print(10.0 / 5.0)
print(10.0 // 5.0)
print(10.0 % 5.0)
print(10.0 * 5.0)
print(10.0 ** 5.0)
```

```python
# string
print("Farm" + "House")
# print("Farm" - "House") -> TypeError: unsupported operand type(s) for -: 'str' and 'str'
print("Farm"*5) # this is repeating a string
```

```python
# int and float - arithmetic operations with different data types
print(10 + 5.0)
print(10 - 5.0)
print(10 / 5.0)
print(10 // 5.0)
print(10 % 5.0)
print(10 * 5.0)
print(10 ** 5.0)
```

```python
# Operator Precedence
# 1. ()
# 2. **
# 3. *, /, //, %
# 4. +, -
# What If Operators Have the Same Precedence? : So Python evaluates them from left to right.

print((4*5)-9+6/7)
print(4%(1+9)**2-60//(7+2))
```
