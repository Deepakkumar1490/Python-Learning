```python
# input() is a built-in Python function used to get information from the user while the program is running.
name = input("Enter you name :")
print(name)
```

```python
# input() Always Returns a String
name = input("Enter you name :")
print(name)
print(type(name))
# Why Does Python Return String? : Because Python doesn't know what the user intends.
```

```python
# Taking an Integer Input
age = int(input("Enter your age : "))
print(age)
print(type(age))
```

```python
# Taking a Float Input
price = float(input("Enter the price :"))
print(price)
print(type(price))
```

```python
# taking multiple input 
name = input("Enter your name: ")
city = input("Enter your city: ")
age = int(input("Enter your age: "))

print("Name:", name)
print("City:", city)
print("Age:", age)
```

```python
# input() + Arithmetic
a = input("Enter a number : ")
b = input("Enter a number : ")
print(a + b)

c = int(input("Enter a number : "))
d = int(input("Enter a number : "))
print(c + d)
```

```python
# input() with split()
# Multiple values : input().split()

numbers = input("Enter numbers: ").split()
print(numbers)
```

```python
# convering numbers to integer
numbers = list(map(int, input("Enter numbers: ").split()))

print(numbers)
```
