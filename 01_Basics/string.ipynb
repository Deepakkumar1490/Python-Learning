```python
# string indexing, index starts from 0
name = "Python"
print(name[0])
print(name[1])
print(name[2])
print(name[3])
print(name[4])
print(name[5])
# print(name[6]) -- IndexError: string index out of range
name[0]
name[1]
```

```python
# negative indexing, from last chararcter it starts with -1
name = "Python"
print(name[-6])
print(name[-5])
print(name[-4])
print(name[-3])
print(name[-2])
print(name[-1])
```

```python
# string slicing, taking a portion of a string
name = "Python"
print(name[0:3])        # The End position is nit included
print(name[0:])
print(name[:3])
print(name[3:])
# The end position is excluded, but if the end position is larger than the string length, 
# Python simply gives you the available characters.
# does not give an error, even though the last valid index is 5.
print(name[:6]) 
print(name[0:10])
```

```python
# Built in string methods
name = "python leaRning"

print(len(name)) # finding length

print(name.upper())

print(name.lower())

print(name.title())     # Coverts Each Word of First character as Capital character

text = "    yes no "
print(text.strip())     # removes spaces from beginning and end

message = "I like Python"
message = message.replace("Python","Java")      # replaces text
print(message)

words = message.split()  # splits a string into pieces
print(words)

result_words = "- ".join(words)     # join multiple strings together
print(result_words)
result_message = "- ".join(message)
print(result_message)
```

```python
# String formatting

# old way
name = "Praveen"
age = 44
print("My name is", name , "and I am", age , "years old")
print(f"My name is {name} and I am {age} years old")
```

```python
# Escape Character

# new line : \n
message = "New\nLine"
print(message)

# new tab : \t
message = "New\ttab"
print(message)

# quote inside a string
message = "Quote Inside a \"String\""
print(message)
```

```python
# Strings are immutable

name = "python"
# name[0] = "J"  : TypeError: 'str' object does not support item assignment
# Python strings cannot be changed after they are created. name[0] can be accessed, 
# but you cannot assign a new value to it.
name = "J" + name[1:] # Create a new string instead of modifying the existing string
print(name)
```
