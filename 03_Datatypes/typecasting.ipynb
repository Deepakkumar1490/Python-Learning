```python
# python in-built functions for typecasting
# int           -> int("10") - 10
# float         -> float("10.5")  - 10.5
# string        -> str(100)   - 100
# boolean       -> bool(1)    - True
# list          -> list("abc") - ['a', 'b', 'c']
# tuple         -> tuple([1,2]) - (1,2)
# set           -> set([1,2,2]) - {1,2}
```

```python
var_str = "100"
print(type(var_str))
print(int(var_str))         # converting string to integer
print("Type of str value is ",type(var_str))
```

```python
var_float = 12.6
print(type(var_float))
print(int(var_float))       # converting float to Integer
# print("Type of float value is " + type(var_float))  -> TypeError: can only concatenate str (not "type") to str
print("Type of float value is",type(var_float))
```

```python
str_float = "2.2"
print(type(str_float))
print(float(str_float))  # converting string to float
```

```python
# print(int(str_float)) ValueError: invalid literal for int() with base 10: '2.2'
# cannot directly convet a string variable which contains the float to integer
# solution : convert a float value inside a string varaiable to integer

str_float = "2.2"
print(int(float(str_float)))
```

```python
# boolean       -> bool("")
print(bool(1))
print(bool(0))
print(bool("Python"))
print(bool(""))
print(bool("True"))
print(bool("False"))
```

```python
print(int(10.9))
print(int(-10.9)) # it will print the value with negative
print(int(+10.9)) # it will not print a '+ ve' character
```

```python
x = "Python"
# print(int(x)) -> ValueError: invalid literal for int() with base 10: 'Python'
```

```python
#   Explicit conversion
#   You tell Python to convert the type.

x = "10"
x = int(x)
print(x)
```

```python
#   Implicit conversion
#   Python automatically converts compatible numeric types in some operations.

x = 10      # int
y = 2.5     # float

result = x + y

print(result)
print(type(result))
```
