```python
# print() is a built-in python function to display information on the screen
# print text
print("Hello World")
print("Welcome")
# Each print function normally displays its content on a new line
```

```python
# print numbers
print(10)
print(10 + 20)
print(25.5)
print(-100)
print("10")
print(type(10))
print(type("10"))
```

```python
# printing variable

name = "Sachin"
age = 53
print(name)
print(age)
```

```python
# print muiltpe values
name = "Sachin"
age = 53
print(name, age) #Python automatically puts a space between the values.
```

```python
# print() can perform calculations
print(10 + 5)
print(10 - 5)
print(10 / 5)
print(10 // 5)
```

```python
# print() with strings vs calculations
print(10 + 20)
print("10 + 20")    # Because Python treats anything inside quotes as text.
print("10" + "20")
```

```python
# sep — separator
# Python uses a space as the separator. -> Internally, print() has: sep=" "
print("A","B","C")
print("A","B","C", sep = " ")
print("A","B","C", sep = "")
print("A","B","C", sep =" | ")
```

```python
# end — what happens after printing?
# By default, print() ends with a newline.
print("A")
print("B")
```

```python
print("Hello", end = " ")
print("World")
```

```python
print("Hello", end = "")
print("World")
```

```python
# sep and end together
print("Python", "Java", "SQL", sep=" | ", end=" END")
```

```python
# Escape characters
# New line : \n
print("Hello\nWorld")
# Tab : \t
print("Hello\tWorld")
# Quotes Inside string
print("Hello World \"Welcome\"")
```

```python
# Printing variables with text
# A common beginner approach
print("My Name is", name, "and", age, "years old")
```

```python
# f-string with print()
print(f"My name is {name} and {age} years old")
a = 10
b = 20
print(f"The sum is {a + b}")

price = 99.26477
print(f"{price : .2f}") # 2f means: Display the number as a floating-point value with 2 decimal places.
print(f"{price : .1f}%")
```

```python
# print() returns None
result = print(10)
print(result) # None : Because print() displays something, but it doesn't return that displayed value
# This distinction between printing and returning becomes extremely important when you learn functions.
```
