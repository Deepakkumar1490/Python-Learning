```python
# len() counts the how many items are inside an object
name = "sachin"
print(len(name))
print(name[len(name) - 1])
age = ""
print(len(age))
age = 12
# print(len(age)) -> TypeError: object of type 'int' has no len()

# If you want to count the digits, you can convert it to a string:
print(len(str(age)))
```

```python
# round() is a Python built-in function used to round a number to a specified number of decimal places.
x = 3.14159
print(round(x))
```

```python
# round(number, ndigits) -> number : The number you want to round, ndigits : How many decimal places you want
print(round(x, 2))
```

```python
print(round(3.2))
print(round(3.8))
print(round(-3.2))
print(round(-3.8))
```

```python
print(round(3.12, 1))
print(round(3.16, 1))
print(round(3.15, 1))
print(round(-3.12, 1))
print(round(-3.16, 1))
print(round(-3.15, 1))
```

```python
x = 12345

print(round(x, -1)) 
print(round(x, -2))
print(round(x, -3))
# ndigits =  2  → decimal places
# ndigits =  1  → decimal places
# ndigits =  0  → whole number
# ndigits = -1  → nearest 10
# ndigits = -2  → nearest 100
# ndigits = -3  → nearest 1000
```

```python
# round() and .5
print(round(1.5))
print(round(2.5))
print(round(3.5))
print(round(4.5))
print(round(6.5))
# When a number is exactly halfway, Python chooses the nearest even number.
# Round-half-to-even helps reduce that bias when rounding many values.
```

```python
# round() returns a value
number = round(3.143)
print(number)
```

```python
# important floating point precision
print(round(3.15, 1))
# Python gives 3.1 because of how floating-point numbers are stored internally.
# Computers don't store most decimal numbers like 3.15 exactly. 
# Python stores a very close binary approximation of 3.15, which is slightly less than the exact value.
# So internally, it is approximately: 3.149999999...
# When rounded to 1 decimal place: 3.149999... → 3.1
```

```python
# abs() means absolute value. It removes the negative sign from a number.
print(abs(10))
print(abs(-10))
print(abs(0))
print(abs(5.5))
print(abs(-5.5))
# Think of it as how far from 0
```

```python
print(abs(-3.7))
print(round(-3.7))
```

```python
temperature = -15
print(temperature)

a = 10
b = 20
print(a - b) # without abs() function
print(abs(a - b)) # with abs() function
```

```python
# abs() with bool

print(abs(True))
print(abs(False))
```

```python
# abs() with complex numbers
x = 3+ 4j # Python supports complex numbers
# Python chose j as the notation for the imaginary part.
print(abs(x))
print(type(x))
# abs() on a complex number does NOT simply remove the negative sign. 
# It calculates the magnitude using the Pythagorean theorem.

# y = 4 + 10k -> SyntaxError: invalid decimal literal
# Python gives a syntax error because 'k' or another letter other than 'J' does not have special meaning for complex numbers.

```

```python
# abs() with type()
x = abs(-13)
print(type(x))

y = abs(-5.5)
print(type(y))

 # z = abs("-13") -> TypeError: bad operand type for abs(): 'str'

z = abs(4 + 10j)
print(type(z))
```
