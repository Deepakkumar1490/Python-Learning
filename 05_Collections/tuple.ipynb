```python
# Tuple Can Store Different Data Types

mixed = (10, "Python", 3.14)
print(mixed)
```

```python
# Indexing

t = ("ban", 10, "can", 100, "tan")
print(t[2])
print(t[3])
print(t[-1])
```

```python
# slicing
t = ("ban", 10, "can", 100, "tan")

print(t[2:])
print(t[1:4])
print(t[:])
print(t[-2:])
```

```python
# concatenation

mixed1 = (10, "Delhi", 3.14)
mixed2 = (50, "Delhi", 5.5)

mixed3 = mixed1 + mixed2
print(mixed3)
```

```python
mixed4 = (10, "Delhi", 3.14)

mixed5 = mixed4 + (20, 30)
print(mixed5)
```

```python
# min, max, sum functions

t = (2,5,7,3)
print(max(t))
print(min(t))
print(sum(t))
```

```python
t = (2,5,7,"Hello")
# print(max(t))  TypeError: '>' not supported between instances of 'str' and 'int'
# print(sum(t)) TypeError: unsupported operand type(s) for +: 'int' and 'str'
```

```python
t = (2,4.5,5,7.8)
print(sum(t))
print(min(t))
print(max(t))
```

```python
# Nested tuples
t = (10, 20, 30, "Hello", (1,2,3))

print(t[4])
print(t[4][1])
print(t[-1][1])
print(t[4][-2])
print(t[-1][-2])
```

```python
# checking a value exits in tuples

t = (10,"hello", 30)
print(10 in t)
print("hello" in t)
print("python" in t)
```

```python
# count() : counts how many time a value occurs
num = (10,20, 30, 10)
print(num.count(10))
print(num.count(20))
```

```python
# index() : finds the first index of value
num = (10,20, 30, 10)
print(num.index(10))
print(num.index(30))
```

```python
# immutablility : means you cannot change an existing tuple

t = (10, 20, 30)
# t[1] = 40 TypeError: 'tuple' object does not support item assignment
```

```python
mixed  = ("welcome", 5, 10, "world")
new_mixed = mixed[0:1] + ("to", "the") + mixed[3:]
print(new_mixed)
```

```python
# Tuple Can Contain Mutable Objects

num = (10, 20, 30, [40, 50])
print(num)
print(type(num))
print(type(num[3]))
# num[3] = [50, 60] TypeError: 'tuple' object does not support item assignment -> cannot change a type element
num[3].append(100) # can modify the list element inside the tuple
print(num)
```

```python
# sort() does NOT work on tuples
# sort() tries to modify the existing collection, but a tuple cannot be modified.
```

```python
# sorted() works with tuples
# sorted() can accept a tuple, but it returns a list.
num = (20, 50, 10, 30)
result = sorted(num)
print(num)
print(result)
print(type(num))
print(type(result))
```

```python
# converts a list to tuple
result = tuple(sorted(num))   
print(result)
```

```python
# descending order
result = sorted(num, reverse=True)
print(result)
```

```python
# packing : putting multiple values into a tuple
person = "a", 2,"c"
print(person)
```

```python
# unpacking
person = ('d', 10, 'c')
name, age, country = person
print(name)
print(age)
print(country)
```

```python
# Unpacking Must Match
person1 = ('d', 10, 'c')
 # name, age = person -> ValueError: too many values to unpack (expected 2, got 3)
```

```python
# Using * During Unpacking -> extended unpacking.
numbers = (10, 20, 30, 40, 50)

first, *middle, last = numbers

print(first)
print(middle)
print(last)
```

```python
# Repetition
num = (1,2)
print(num*3)
```
