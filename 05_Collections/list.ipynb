```python
# a list is a collection of muliple values stored together in a single variable
num = [10, 20, 30, 40, 50]
print(num)
print(type(num)) # list of int
```

```python
char = ["A", "B", "C"]
print(char)
print(type(char)) # list of string
```

```python
mixed = [10,"Python", 3.14, True]
print(mixed)
print(type(mixed)   # list can contain different data types
```

```python
# indexing 
num = [10, 20, 30, 50]
print(num[0])
print(num[2])
print(num[-1])
print(num[-3])
```

```python
# mutable : can change the content of the list after creating it
mixed = [10,"Python", 3.14, True]
mixed[1] = "Java"
print(mixed)

# mixed[6] = 60 IndexError: list assignment index out of range
```

```python
# length : len()
mixed = [10,"Python", 3.14, True]
print(len(mixed))
```

```python
# nested list : list can contain another list
mixed = [10,"Python", 3.14, True, ["a", "b", "c"]]
print(mixed)
print(len(mixed))
print(mixed[4])
print(mixed[4][0])
print(mixed[4][1])
print(mixed[4][2])
print(mixed[-1][0])
print(mixed[-1][-1])
print(mixed[-1][-3])
```

```python
# list concat
# list added is appendent with element
list1 = [1,2,3,4]
list2 = [5,6]
new_list = list1 + list2
print(new_list)

list3 = list1 + [5,6]
print(list3)
```

```python
# checking whether an Item exits

list = [1,2,3,4,5]
print(1 in list)
print(4 in list)
print(10 in list)

char = ["a", "b", "c", "d"]
# print(a in char) NameError: name 'a' is not defined
print("a" in char)
print("d" in char)
print("z" in char)
```

```python
# append() : adds an one object at the end

list = [1,2,3]
list.append([4,5])
print(list)

char = ["a", "b", "c"]
char.append("d")
print(char)
```

```python
# insert() : allows you to add an item at a specific position
mixed = [10,"Python", 3.14, True]
mixed.insert(1, 20)
mixed.insert(10, 30) # inserts value at last of list
print(mixed)
```

```python
# remove() : removes a specific value at the first occuring value
mixed = [10,"Python", 3.14, True]
mixed.remove(3.14)
print(mixed)

num = [10, 20, 10, 10]
num.remove(10)
print(num)
```

```python
# pop() : removes an item using its index. By default, it deletes the last object from the list

num = [10, 20, 30, 40, 50]
num.pop()
print(num)

num.pop(1)
print(num)
```

```python
# del : can also delete the element using an index

num = [10, 20, 30, 40, 50]
del num[2]
# del num[10] IndexError: list assignment index out of range
print(num)
```

```python
# extend() : adds multiple items to a list

num = [10, 20, 30, 40, 50]
num.extend([60,70])
print(num)
```

```python
# sort() : By default, it sorts ascending order. It does not return value, it just sorts the value

num = [20, 40, 10, 60, 50]
num1 = num.sort()
print(num)
print(num1)
print(num.sort())
```

```python
# descending order 
num = [20, 40, 10, 60, 50]
num.sort(reverse=True)
print(num)
```

```python
# sorted() : creates/returns a sorted result while leaving the original list unchanged

a = [20, 30, 10, 60, 50]
b = sorted(a)

print(a)
print(b)
print(sorted(a))
```

```python
# list slicing : taking part of a list
num = [10, 20, 30, 40, 50]
print(num[1:4])
print(num[:4])
print(num[1:])
print(num[:])
```

```python
# reverse() : reverse the current order, does not mean sorting
num = [30, 10, 60, 40]
num.reverse()
print(num)
```

```python
# count() : it tells how many times the value occurs
num = [10, 20, 10, 50, 10, 70]
print(num.count(10))
print(num.count(50))
```

```python
# index() : it return the index of the first matching value

num = [10, 20, 10, 50, 10, 70]
print(num.index(10))
print(num.index(50))
```

```python
# shallow copy : 
a = [10, 20, 30]
b = a

print(a)
print(b)

a[2] = 40  # both a and varaible refers to the same list
print(a)
print(b)

num1 = [10, 20, 30]
num2 = num1.copy()

num2[2] = 100
print(num1)
print(num2)

num3 = [10, 20, 30]
num4 = num3[:]

num3[2] = 100
print(num3)
print(num4)
```
